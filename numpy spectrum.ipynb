```python
import numpy as np
wavelength = np.array([1549.8,1549.9,1550.0,1550.1,1550.2])
intensity = np.array([0.92,0.71,0.45,0.69,0.90])
print('波长元素类型：',wavelength.dtype)
print('波长数据维度：',wavelength.ndim)
print('波长数据形状：',wavelength.shape)
print('波长元素数量：',wavelength.size)
print('波长元素占用字节：',wavelength.itemsize)
print('波长元素总字节数：',wavelength.nbytes)
print('波长数据类型：',type(wavelength))
```

```python
spectra = np.array([
    [0.92, 0.71, 0.45, 0.69, 0.90],
    [0.93, 0.76, 0.51, 0.65, 0.88],
    [0.91, 0.80, 0.59, 0.61, 0.86]
]) #行是指不同氢气浓度下的光谱  列是指不同波长采样点的光强
print('波长元素类型：',spectra.dtype)
print('波长数据维度：',spectra.ndim)
print('波长数据形状：',spectra.shape)
print('波长元素数量：',spectra.size)
print('波长元素总字节数：',spectra.nbytes)
```

# 光谱数据报告

```python
import numpy as np
#一条光谱的波长和光强
wavelength = np.array([1549.8,1549.9,1550.0,1550.1,1550.2])
intensity = np.array([0.92,0.71,0.45,0.69,0.90])
#三种氢气浓度下的光强，每行是一条光谱
spectra = np.array([
    [0.92, 0.71, 0.45, 0.69, 0.90],
    [0.93, 0.76, 0.51, 0.65, 0.88],
    [0.91, 0.80, 0.59, 0.61, 0.86]
])
print('=======光谱数据报告=======')
print("波长数组维度：", wavelength.ndim)
print("波长数组形状：", wavelength.shape)
print("光强数组形状：", intensity.shape)
print("光谱采样点数：", wavelength.size)
print("多浓度光谱数量：", spectra.shape[0])
print("每条光谱采样点数：", spectra.shape[1])
print("数据类型：", spectra.dtype)
print("多浓度光强数组占用空间：", spectra.nbytes, "bytes")
```

# 创建波长数组

```python
wavelength = np.linspace(1520,1580,2001,dtype = int)
print(wavelength)
print(wavelength.shape)
print(wavelength.size)
print(wavelength.dtype)
space = (1580 - 1520)/(2001 - 1)
print(f'{space}nm')
```

## 比较两种创建方式

```python
wave1 = np.linspace(1520,1580,7, dtype = int)
wave2 = np.arange(1520,1580,10)
print(wave1)
print(wave2)
print(wave1.size)
print(wave2.size)
```
