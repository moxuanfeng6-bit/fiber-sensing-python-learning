# 光谱寻谷与误差分析

```python
import numpy as np
rng = np.random.default_rng(42)
wavelength = np.linspace(1520, 1580, 2001)
true_center = 1550.25

# 模拟光谱：中心附近光强降低
clean_intensity = 1 - 0.6 * np.exp(
    -((wavelength - true_center) / 0.8) ** 2
)

# 添加随机噪声
noise = rng.normal(0, 0.015, size=wavelength.size)
measured_intensity = clean_intensity + noise
```

```python
#输出波长、光强的形状和类型
print(wavelength.shape,wavelength.dtype)
print(measured_intensity.shape,measured_intensity.dtype)
```

```python
#截取1545～1555 nm的局部光谱
mask = (wavelength >= 1545) & (wavelength <= 1555)  #布尔索引划范围，判断 T or F
selected_wavelength = wavelength[mask]
selected_intensity = measured_intensity[mask]
print(selected_wavelength)
print(selected_intensity)
```

```python
#计算局部光强的均值、标准差和最小值
print(f'{np.mean(selected_intensity):.2f}')
print(f'{np.std(selected_intensity):.2f}')
print(f'{np.min(selected_intensity):.2f}')
```

```python
#找到局部光谱最小光强的下标
print(np.argmin(selected_intensity))
```

```python
#根据下标取出对应的波长
min_index = np.argmin(selected_intensity)
print(selected_wavelength[min_index])
print(min_index)
```

```python
#计算检测波长与真实中心的误差
detected_wavelength = selected_wavelength[min_index]
error_nm = detected_wavelength - true_center
absoltue_error_nm = abs(error_nm)
print(error_nm,'nm')
print(absoltue_error_nm,'nm')
```

```python
#把误差从nm换成pm
absoltue_error_pm = absoltue_error_nm * 1000
print(absoltue_error_pm,'pm')
```

```python
#找到光强最低的5个点，输出对应波长
sorted_indices = np.argsort(selected_intensity)
lowest_indices = sorted_indices[:5]
print(selected_wavelength[lowest_indices])
print(selected_intensity[lowest_indices])
```

# 传感器标定曲线

```python
concentrations = np.array([0, 0.5, 1.0, 1.5, 2.0])

# 浓度每增加1个百分点，波长增加0.08 nm
theoretical_centers = 1550.25 + 0.08 * concentrations

# 测量偏差，单位nm
offsets = np.array([0.002, -0.003, 0.001, 0.004, -0.002])
measured_centers = theoretical_centers + offsets

theoretical_shifts = theoretical_centers - theoretical_centers[0]
measured_shifts = measured_centers - measured_centers[0]
```

```python
plt.figure(figsize=(10,5))
plt.plot(concentrations,theoretical_shifts,color = 'blue',label = 'Theoretical')
plt.scatter(concentrations,measured_shifts,color = 'red',label = 'Measured',s=20,zorder=3)
plt.title('Simulated Calibration Curve')
plt.xlabel('Hydrogen concentration (%)')
plt.ylabel('Wavelength shift (nm)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

# 分别提取慢信号和快信号

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.sans-serif'] = 'Microsoft YaHei'
from scipy import fftpack

N = 1000
f_s = 100

t = np.arange(N) / f_s

slow = np.sin(2*np.pi*2*t)
fast = 0.5 * np.sin(2*np.pi*15*t)
mixed = slow + fast
```

```python
F = fftpack.fft(mixed)
freq = fftpack.fftfreq(N, 1/f_s)

F_low = F*(np.abs(freq)<5) #低通滤波器
F_high = F*(np.abs(freq)>5) #高通滤波器

slow_recovered = fftpack.ifft(F_low).real
fast_recovered = fftpack.ifft(F_high).real

slow_error = np.max(np.abs(slow_recovered - slow))
fast_error = np.max(np.abs(fast_recovered - fast))

print(f'慢信号最大绝对误差：{slow_error:.3e}')
print(f'快信号最大绝对误差：{fast_error:.3e}')
```
